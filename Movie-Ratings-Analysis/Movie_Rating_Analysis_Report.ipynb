```python
df.drop_duplicates(inplace=True)
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print(np.mean(df["Rating"]))
```

```python
print(np.median(df["Rating"]))
```

```python
print(np.std(df["Rating"]))
```

```python
print(np.mean(df["Budget"]))
print(np.median(df["Budget"]))
print(np.std(df["Budget"]))
```

```python
print(np.mean(df["Revenue"]))
print(np.median(df["Revenue"]))
print(np.std(df["Revenue"]))
```

```python
genre_analysis = df.groupby("Genre")["Rating"].mean()

genre_analysis
```

```python
director_analysis = df.groupby("Director")["Rating"].mean()

director_analysis
```

```python
year_analysis = df.groupby("Year")["Rating"].mean()

year_analysis
```

```python
genre_analysis.plot(kind="bar", figsize=(8,5))

plt.title("Average Rating by Genre")

plt.xlabel("Genre")

plt.ylabel("Average Rating")

plt.show()
```

```python
director_analysis.plot(kind="bar", figsize=(10,5))

plt.title("Average Rating by Director")

plt.xlabel("Director")

plt.ylabel("Average Rating")

plt.xticks(rotation=45)

plt.show()
```

```python
year_analysis.plot(marker="o")

plt.title("Average Ratings Over Years")

plt.xlabel("Year")

plt.ylabel("Average Rating")

plt.grid(True)

plt.show()
```

```python
plt.scatter(df["Budget"], df["Rating"])

plt.title("Budget vs Rating")

plt.xlabel("Budget")

plt.ylabel("Rating")

plt.show()
```

```python
plt.scatter(df["Revenue"], df["Rating"])

plt.title("Revenue vs Rating")

plt.xlabel("Revenue")

plt.ylabel("Rating")

plt.show()
```

```python
top_movies = df.sort_values("Rating", ascending=False).head(10)

plt.barh(top_movies["Movie"], top_movies["Rating"])

plt.title("Top Rated Movies")

plt.xlabel("Rating")

plt.show()
```
